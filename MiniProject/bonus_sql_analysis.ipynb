```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("cleaned_ai_learning_lab.csv")
```

```python
df["total_score"] = (
    df["assignment_score"] +
    df["quiz_score"]
)
```

```python
conn = sqlite3.connect(
    "learning_lab.db"
)
```

```python
df.to_sql(
    "student_learning",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query1 = """
SELECT topic,
       AVG(assignment_score)
FROM student_learning
GROUP BY topic;
"""

result1 = pd.read_sql_query(
    query1,
    conn
)

print(result1)
```

```python
query2 = """
SELECT batch,
       COUNT(*)
FROM student_learning
GROUP BY batch;
"""

result2 = pd.read_sql_query(
    query2,
    conn
)

print(result2)
```

```python
query3 = """
SELECT student_name,
       total_score
FROM student_learning
WHERE total_score < 80;
"""

result3 = pd.read_sql_query(
    query3,
    conn
)

print(result3)
```

```python
conn.close()
```
