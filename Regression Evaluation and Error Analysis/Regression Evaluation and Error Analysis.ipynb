3. Analyze how increasing alpha affects the Ridge Regression model.
4. Recommend the final model training and evaluation strategy.
5. Justify whether **MAE** or **MSE/RMSE** should be used when the business wants to minimize large vehicle price prediction errors.
6. Explain one important difference between **Ridge Regression** and **Lasso Regression**.

## Expected Result

The analysis should conclude that:

- **Best Alpha = 1**
- **Highest Mean R² = 0.682**
- **Lowest Mean MAE = 219400**

The final recommendation should suggest using:

- Ridge Regression with **alpha = 1**
- Retraining on the complete training dataset
- Evaluating on the test dataset
- Using **RMSE (or MSE)** when large prediction errors are more costly
- Understanding that Ridge shrinks coefficients while Lasso can reduce coefficients to exactly zero

```python
import pandas as pd

# Cross-validation results
cv_results = pd.DataFrame({
    "Alpha": [0, 0.001, 0.01, 0.1, 1, 10, 50, 100, 200, 500, 1000, 5000, 10000],
    "Mean R2": [0.671, 0.671, 0.672, 0.675, 0.682, 0.679, 0.668, 0.651, 0.630, 0.588, 0.541, 0.412, 0.301],
    "Mean MAE": [221500, 221490, 221350, 220800, 219400, 220100, 222800, 226500, 231200, 241000, 252300, 278900, 309400]
})

print("Cross Validation Results")
display(cv_results)
```

```python
best_model = cv_results.loc[cv_results["Mean R2"].idxmax()]

print("Best Alpha Found")
print("---------------------------")
print(f"Alpha      : {best_model['Alpha']}")
print(f"Mean R²    : {best_model['Mean R2']}")
print(f"Mean MAE   : {best_model['Mean MAE']}")
```

```python
print("Alpha Behaviour Analysis")
print("---------------------------")

print("""
• Alpha = 0 means no regularization.

• From Alpha 0 to Alpha 1:
  - Mean R² increases.
  - Mean MAE decreases.
  - Model performance improves.

• Beyond Alpha = 1:
  - Mean R² decreases.
  - Mean MAE increases.
  - Strong regularization causes underfitting.

• Very large alpha values
  (1000, 5000, 10000)
  produce poor predictions.
""")
```

```python
print("Recommended Training Strategy")
print("---------------------------")

print("""
1. Select Alpha = 1.

2. Retrain Ridge Regression
   using the complete training dataset.

3. Evaluate the trained model
   on the unseen test dataset.

4. Report evaluation metrics
   such as MAE, RMSE and R².
""")
```

```python
print("Recommended Error Metric")
print("---------------------------")

print("""
Business Goal:
Reduce large vehicle price prediction errors.

Recommended Metric:
RMSE (Root Mean Squared Error)

Reason:
RMSE squares the prediction errors,
therefore large mistakes receive a
much larger penalty than MAE.

Hence RMSE is preferred when
large pricing mistakes are costly.
""")
```

```python
print("Ridge vs Lasso")
print("---------------------------")

print("""
Ridge Regression
----------------
• Uses L2 Regularization.
• Shrinks coefficients.
• Does NOT make coefficients exactly zero.
• Keeps all features.

Lasso Regression
----------------
• Uses L1 Regularization.
• Shrinks coefficients.
• Can reduce coefficients exactly to zero.
• Performs automatic feature selection.
""")
```

```python
print("=" * 50)
print("FINAL CONCLUSION")
print("=" * 50)

print(f"Best Alpha : {best_model['Alpha']}")
print(f"Highest Mean R² : {best_model['Mean R2']}")
print(f"Lowest Mean MAE : {best_model['Mean MAE']}")

print("""
Final Recommendation

✔ Use Ridge Regression

✔ Choose Alpha = 1

✔ Retrain the model on the complete
  training dataset.

✔ Evaluate on the unseen test data.

✔ Use RMSE when the business wants
  to minimize large pricing errors.
""")
```
