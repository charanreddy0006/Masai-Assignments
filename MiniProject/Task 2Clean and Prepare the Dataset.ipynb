```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("ai_learning_lab.csv")
```

```python
df = df.drop_duplicates()
```

```python
df["student_name"] = df["student_name"].str.strip()
df["topic"] = df["topic"].str.strip()
df["tool_used"] = df["tool_used"].str.strip()
```

```python
df["attendance"] = df["attendance"].str.title()
df["lab_completed"] = df["lab_completed"].str.title()
df["api_used"] = df["api_used"].str.title()
```

```python
df.loc[
    (df["assignment_score"] < 0) |
    (df["assignment_score"] > 100),
    "assignment_score"
] = np.nan
```

```python
df.loc[
    (df["quiz_score"] < 0) |
    (df["quiz_score"] > 100),
    "quiz_score"
] = np.nan
```

```python
df.loc[
    df["study_hours"] < 0,
    "study_hours"
] = np.nan
```

```python
df.loc[
    (df["feedback_rating"] < 1) |
    (df["feedback_rating"] > 5),
    "feedback_rating"
] = np.nan
```

```python
mean_assignment = df["assignment_score"].mean()

df["assignment_score"] = df["assignment_score"].fillna(
    mean_assignment
)
```

```python
mean_quiz = df["quiz_score"].mean()

df["quiz_score"] = df["quiz_score"].fillna(
    mean_quiz
)
```

```python
median_hours = df["study_hours"].median()

df["study_hours"] = df["study_hours"].fillna(
    median_hours
)
```

```python
median_rating = df["feedback_rating"].median()

df["feedback_rating"] = df["feedback_rating"].fillna(
    median_rating
)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
df.to_csv(
    "cleaned_ai_learning_lab.csv",
    index=False
)
```
