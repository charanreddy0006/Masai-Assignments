```python
import pandas as pd

df = pd.read_csv("cleaned_ai_learning_lab.csv")
```

```python
df["total_score"] = (
    df["assignment_score"] +
    df["quiz_score"]
)
```

```python
def performance(score):

    if score >= 160:
        return "Excellent"

    elif score >= 120:
        return "Good"

    elif score >= 80:
        return "Average"

    else:
        return "Needs Support"


df["performance_level"] = df["total_score"].apply(
    performance
)  
```

```python
print(df["assignment_score"].mean())
```

```python
print(df["quiz_score"].mean())
```

```python
print(df["study_hours"].mean())
```

```python
print(
    df["performance_level"]
    .value_counts()
)
```

```python
topic_avg = df.groupby(
    "topic"
)["total_score"].mean()

print(topic_avg)
```

```python
batch_avg = df.groupby(
    "batch"
)["total_score"].mean()

print(batch_avg)
```

```python
needs_support = df[
    df["performance_level"]
    == "Needs Support"
]

print(needs_support)
```
