```python
import pandas as pd
import matplotlib.pyplot as plt
import json
```

```python
df = pd.read_csv("cleaned_ai_learning_lab.csv")
```

```python
df["total_score"] = (
    df["assignment_score"] +
    df["quiz_score"]
)
```

```python
topic_scores = (
    df.groupby("topic")
    ["total_score"]
    .mean()
)

topic_scores.plot(kind="bar")

plt.title("Average Total Score by Topic")
plt.xlabel("Topic")
plt.ylabel("Average Total Score")

plt.tight_layout()

plt.savefig("topic_score_chart.png")
plt.close()
```

```python
def performance(score):

    if score >= 160:
        return "Excellent"

    elif score >= 120:
        return "Good"

    elif score >= 80:
        return "Average"

    else:
        return "Needs Support"


df["performance_level"] = (
    df["total_score"]
    .apply(performance)
)

performance_counts = (
    df["performance_level"]
    .value_counts()
)

performance_counts.plot(kind="bar")

plt.title("Student Count by Performance Level")
plt.xlabel("Performance Level")
plt.ylabel("Student Count")

plt.tight_layout()

plt.savefig(
    "performance_level_chart.png"
)

plt.close()
```

```python
attendance_counts = (
    df["attendance"]
    .value_counts()
)

attendance_counts.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Attendance Distribution")

plt.ylabel("")

plt.savefig(
    "attendance_chart.png"
)

plt.close()
```

```python
study_hours = (
    df.groupby("topic")
    ["study_hours"]
    .mean()
)

study_hours.plot(kind="line", marker="o")

plt.title("Average Study Hours by Topic")

plt.xlabel("Topic")

plt.ylabel("Average Study Hours")

plt.tight_layout()

plt.savefig(
    "study_hours_chart.png"
)

plt.close()
```

```python
summary = {

    "total_students":
        int(len(df)),

    "average_assignment_score":
        float(
            df["assignment_score"]
            .mean()
        ),

    "average_quiz_score":
        float(
            df["quiz_score"]
            .mean()
        ),

    "average_study_hours":
        float(
            df["study_hours"]
            .mean()
        ),

    "most_common_topic":
        df["topic"]
        .mode()[0]
}
```

```python
with open(
    "learning_summary.json",
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )
```
