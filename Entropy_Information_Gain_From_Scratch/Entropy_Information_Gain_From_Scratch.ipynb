# What is Entropy?

Entropy measures the impurity or uncertainty in a dataset.

- High entropy → Mixed classes
- Low entropy → Pure classes

If all records belong to one class,

Entropy = 0

If classes are equally mixed,

Entropy is maximum.

# Entropy Formula

\[
Entropy(S) = -\sum p_i \log_2(p_i)
\]

Where:

- p = Probability of each class

# What is Information Gain?

Information Gain measures how much uncertainty decreases after splitting the dataset.

Decision Trees always choose the feature with the highest Information Gain.

# Information Gain Formula

\[
IG = ParentEntropy - WeightedChildEntropy
\]

Higher Information Gain means a better feature for splitting.

# Import Required Library

We only use Python's standard library.

Required module:

- math

```python
import math 
```

```python
#Create Dataset
record_list = [
    {"Outlook": "Sunny", "Humidity": "High", "PlayTennis": "No"},
    {"Outlook": "Sunny", "Humidity": "High", "PlayTennis": "No"},
    {"Outlook": "Overcast", "Humidity": "High", "PlayTennis": "Yes"},
    {"Outlook": "Rainy", "Humidity": "High", "PlayTennis": "Yes"},
    {"Outlook": "Rainy", "Humidity": "Normal", "PlayTennis": "Yes"},
    {"Outlook": "Rainy", "Humidity": "Normal", "PlayTennis": "No"},
    {"Outlook": "Overcast", "Humidity": "Normal", "PlayTennis": "Yes"},
]
```

# Compute Entropy Function

This function:

- Counts class labels
- Computes probabilities
- Applies the entropy formula

```python
def compute_entropy(labels):

    if len(labels) == 0:
        return 0.0

    counts = {}

    for label in labels:
        counts[label] = counts.get(label, 0) + 1

    if len(counts) == 1:
        return 0.0

    entropy = 0

    total = len(labels)

    for count in counts.values():

        probability = count / total

        if probability > 0:
            entropy -= probability * math.log2(probability)

    return round(entropy, 4)
```

# Compute Information Gain Function

This function:

- Computes Parent Entropy
- Splits the dataset
- Computes Child Entropy
- Returns Information Gain

```python
def compute_information_gain(records, feature, target):

    parent_labels = [row[target] for row in records]

    parent_entropy = compute_entropy(parent_labels)

    groups = {}

    for row in records:
        groups.setdefault(row[feature], []).append(row)

    weighted_entropy = 0

    total = len(records)

    for subset in groups.values():

        subset_labels = [row[target] for row in subset]

        weighted_entropy += (
            len(subset) / total
        ) * compute_entropy(subset_labels)

    information_gain = parent_entropy - weighted_entropy

    return round(information_gain, 4)
```

```python
#Calculate Entropy
labels = [
    row["PlayTennis"]
    for row in record_list
]

entropy = compute_entropy(labels)

print("Entropy:", entropy)
```

```python
# Calculate Information Gain
ig = compute_information_gain(
    record_list,
    "Outlook",
    "PlayTennis"
)

print("Information Gain:", ig)
```
