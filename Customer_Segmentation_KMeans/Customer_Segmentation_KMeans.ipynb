# Customer Segmentation using K-Means Clustering

## Objective

This project demonstrates how to segment customers based on their **Annual Income** and **Spending Score** using the K-Means clustering algorithm.

The workflow includes:

- Data preparation
- Feature scaling
- Selecting the optimal number of clusters using Silhouette Score
- Training the final K-Means model
- Creating customer segments
- Analyzing cluster characteristics

# Problem Statement

A retail analytics team wants to group customers based on purchasing behavior.

Using Annual Income and Spending Score, we will:

- Scale the features
- Evaluate multiple K values
- Select the best K using Silhouette Score
- Train the final K-Means model
- Analyze the resulting customer segments

# What is K-Means Clustering?

K-Means is an unsupervised machine learning algorithm used to divide similar data points into groups called clusters.

The algorithm works by:

1. Selecting K cluster centers.
2. Assigning each point to its nearest center.
3. Updating the cluster centers.
4. Repeating until convergence.

The goal is to minimize the distance between points and their cluster centers.

# Why Feature Scaling?

K-Means is a distance-based algorithm.

If one feature has a much larger scale than another, it will dominate the distance calculation.

StandardScaler standardizes every feature to have:

- Mean = 0
- Standard Deviation = 1

This ensures both features contribute equally during clustering.

# What is Silhouette Score?

Silhouette Score measures the quality of clustering.

Range:

- -1 → Poor clustering
- 0 → Overlapping clusters
- +1 → Well-separated clusters

Higher Silhouette Score indicates better clustering.

```python
import pandas as pd

from sklearn.cluster import KMeans

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import silhouette_score
```

```python
# Create Dataset
data = [
    {"CustomerID":1,"AnnualIncome":15,"SpendingScore":39},
    {"CustomerID":2,"AnnualIncome":16,"SpendingScore":81},
    {"CustomerID":3,"AnnualIncome":17,"SpendingScore":6},
    {"CustomerID":4,"AnnualIncome":18,"SpendingScore":77},
    {"CustomerID":5,"AnnualIncome":19,"SpendingScore":40},
    {"CustomerID":6,"AnnualIncome":62,"SpendingScore":55},
    {"CustomerID":7,"AnnualIncome":65,"SpendingScore":63},
    {"CustomerID":8,"AnnualIncome":70,"SpendingScore":14},
    {"CustomerID":9,"AnnualIncome":75,"SpendingScore":85},
    {"CustomerID":10,"AnnualIncome":78,"SpendingScore":20},
]

df = pd.DataFrame(data)

df.head()
```

```python
# Select Features
X = df[
    [
        "AnnualIncome",
        "SpendingScore"
    ]
]
```

```python
# Feature Scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

# Find the Best Number of Clusters

We evaluate K values:

- K = 2
- K = 3
- K = 4

using Silhouette Score.

```python
scores = {}

for k in [2,3,4]:

    model = KMeans(

        n_clusters=k,

        random_state=42,

        n_init="auto"
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    scores[k] = score

    print(f"K={k}  Silhouette Score={score:.4f}")
```

```python
# Select the Best K
best_k = max(
    scores,
    key=scores.get
)

print("Best K =", best_k)
```

```python
# Train Final K-Means Model
final_model = KMeans(

    n_clusters=best_k,

    random_state=42,

    n_init="auto"
)

df["Cluster"] = final_model.fit_predict(
    X_scaled
)

df
```

```python
# Cluster Profile Analysis
cluster_profile = (

    df.groupby("Cluster")[
        [
            "AnnualIncome",
            "SpendingScore"
        ]
    ]

    .mean()

    .round(2)

)

print(cluster_profile)
```

# Conclusion

The optimal number of customer segments is selected using the highest Silhouette Score.

Each cluster represents customers with similar income and spending patterns, enabling businesses to design targeted marketing strategies.
